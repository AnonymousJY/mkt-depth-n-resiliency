```python
import pandas as pd
```

```python
df = pd.read_csv('/Users/josephyi/Downloads/csv_files/throwsh.csv')
```

```python
# df['date'] = pd.to_datetime(df['date'], format='%Y-%m-%d')
idx = df.groupby(df['date'].dt.year)['vol'].idxmax()
nested_df1 = df.loc[idx, ['date', 'vol']].reset_index(drop=True)
nested_df1['date'] = nested_df1['date'].astype(str)
nested_df1
```

```python
max_close = df.groupby(df['date'].dt.year)['close'].transform('max')
nested_df2 = df.loc[df['close']==max_close, ['date', 'close']].reset_index(drop=True)
```

```python

def df_str(df):
    return df.to_string(index=True)

nested = [
    [nested_df1, nested_df2],
]

pretty = [[df_str(a), df_str(b)] for a, b in nested]
nested
```
